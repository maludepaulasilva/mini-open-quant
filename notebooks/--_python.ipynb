```python
import yfinance as yf
import pandas as pd
```

```python
df = yf.download("PETR4.SA", start="2020-01-01", auto_adjust=True)
precos = df["Close"]

ret = precos.pct_change().dropna()
```

```python
print("Retorno médio diário:", ret.mean())
print("Desvio-padrão diário:", ret.std())

precos.plot(title="PETR4")
```
